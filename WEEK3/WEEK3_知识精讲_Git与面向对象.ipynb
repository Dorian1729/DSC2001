## 1. Git 与 GitHub

Git 是本地使用的分布式版本控制工具；GitHub 是托管 Git 仓库、便于协作的平台。GitHub Desktop 为 Git/GitHub 操作提供图形界面。区分工具与托管平台，有助于理解为什么本地提交后还需要 push 才能上传到 GitHub。

| 名称 | 讲义中的含义 |
|---|---|
| Working directory | 项目文件所在的工作目录 |
| Repository | 保存项目正式历史记录的仓库 |
| Commit | 在本地保存一份变更快照 |
| Push | 把本地提交上传到远程仓库 |
| Pull | 把远程仓库的更新合并到当前分支 |

## 2. 常见仓库操作

GitHub Desktop 安装后可登录 GitHub，并在设置中配置外部编辑器和提交作者姓名、邮箱。新建仓库时选择名称和本地路径，可按需要初始化 README、Git ignore 和 License；Clone 是把远程仓库复制到本地。修改文件后，Changes 页面会显示变更；选择要提交的文件、填写说明并 Commit，才会形成一个本地快照。Push origin 将提交上传到 GitHub。History 页面可查看过去提交及其标识。

命令行流程对应同一组概念：`git init` 初始化仓库，`git status` 查看状态，`git add` 把变更放入 staging area，`git commit` 保存快照。提交前先 add，因此“文件已修改”和“修改已准备提交”是两个状态。

```bash
git init
git status
git add demo1.py
git commit -m "Add demo file"
git log
```

配置 Git 用户名和邮箱用于提交作者信息。添加远程仓库后，用 push 上传分支；pull 获取远程更新。

```bash
git config --global user.name "MyName"
git config --global user.email "my.name@example.com"
git remote add origin https://github.com/username/repository.git
git push -u origin master
git pull origin master
```

## 3. 分支、合并与协作

分支是独立的开发线，可让功能或修复在不直接改动主线的情况下完成。创建、切换分支后，在该分支上修改并提交；完成后可合并回 main/master。GitHub Desktop 的流程是先切换目标主分支，再选择 Merge into current branch 并指定来源分支。

```bash
git checkout -b branch1
git add .
git commit -m "Update demo"
git switch master
git merge branch1
```

若两个分支改动了同一文件的同一部分，合并可能出现冲突。打开冲突文件，决定保留哪一边或如何组合，保存后再 add 并提交合并结果。Pull request 则用于在 GitHub 上分享已 push 的分支变更，供协作者查看和合并。

## 4. 撤销变更时区分状态

讲义按变更是否提交、是否 push 区分处理方式：
- 尚未提交：可在 GitHub Desktop 丢弃文件变更；命令行示例用 `git checkout -- filename`。
- 已提交但未 push：可 Undo/Reset，让变更回到工作区以便修改；Reset 会影响之后的提交。
- 已 push：用 Revert 生成一个新提交来反向撤销。讲义提醒不要对已共享提交使用会改写历史的 Undo/Reset。
- 命令行示例：`git revert <commit-hash>`，或用 `git revert HEAD` 撤销最近提交。

## 5. 从重复代码到对象

OOP 讲义从“计算两个数的平方和”开始：重复步骤可以先提取成函数；函数通过参数接收输入，并用返回值交回结果，避免把结果放在会被多次调用改写的全局变量里。若接下来还要支持立方和、四次方和，继续为每种幂次写新函数会增加管理负担。

````python
def sum_squares(a, b):
    return a ** 2 + b ** 2

def sum_cubes(a, b):
    return a ** 3 + b ** 3

print(sum_squares(2, 3), sum_cubes(2, 3))
````

讲义的面向对象方案把幂次作为对象内部状态，把运算组织成方法。创建不同幂次的对象后，对外仍调用同一个 `calculate` 操作。

````python
class SumPower:
    def __init__(self, power_order):
        self.power_order = power_order

    def power(self, value):
        return value ** self.power_order

    def calculate(self, x, y):
        return self.power(x) + self.power(y)

sum_squares = SumPower(2)
sum_cubes = SumPower(3)
print(sum_squares.calculate(2, 3))
print(sum_cubes.calculate(2, 3))
````

## 6. OOP 概念

- **封装**：把属性和方法组织到类中，隐藏内部细节，通过少量公开方法（API）让外部使用；必要时提供 get/set 方法。
- **继承**：子类沿用父类的属性和方法，并能增加功能。Python 中可用 `super()` 调用父类构造方法。
- **多态**：同一个方法名可以由不同对象以不同方式实现；子类覆盖父类方法后，调用者仍可使用共同的方法名。
- **抽象**：只呈现必要操作，把实现细节交给具体子类。讲义以抽象类和抽象方法说明这一点。

讲义用 `ABC` 和 `@abstractmethod` 表达 Python 抽象类。抽象类可以包含已实现方法；抽象方法声明子类必须提供的操作。材料还比较抽象类与接口：抽象类是相关对象的部分实现蓝图，接口表达可由不同类实现的能力约定；Python 没有 Java 式的 `interface` 关键字。

````python
from abc import ABC, abstractmethod

class Killer(ABC):
    @abstractmethod
    def kill(self, target_name):
        pass

class GunKiller(Killer):
    def kill(self, target_name):
        print(f"I kill {target_name} by guns")

GunKiller().kill("target")
````

### Python 中的封装命名约定

OOP source 说明，Python 不用严格的 private/protected 关键字控制访问，而是用命名约定表达意图：普通名称是公开属性；单下划线（如 `_age`）提示这是内部使用；双下划线（如 `__salary`）会触发名称改写，减少意外的直接访问或子类覆盖。讲义通过公开方法提供对内部数据的访问。

讲义还区分 merge 与 rebase：merge 会形成带有两条父提交线的合并提交；rebase 将分支提交接到目标分支后面，使提交历史呈现为一条线。两者都是把分支工作整合到目标分支的方式。

### GitHub Desktop 与命令行操作补充

GitHub Desktop 设置还包括界面外观；提交历史显示提交标识（讲义示例为 7 位字符）。命令行演示从 git --version、git help 开始；设置用户名和邮箱后，按 status → add → commit 查看并保存变更。远程仓库的 push/pull 对应上传与获取更新。命令行合并冲突同样需要编辑冲突文件、add 后提交合并。

### 用继承扩展幂次求和类

讲义还示范子类调用父类构造方法，并新增公开操作打印内部幂次。

````python
class SumPowerWithPrint(SumPower):
    def __init__(self, power_order):
        super().__init__(power_order)

    def print_power_order(self):
        print(self.power_order)

sum_fourth = SumPowerWithPrint(4)
sum_fourth.print_power_order()
````
