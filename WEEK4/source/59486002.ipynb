1. Write a function to check if a given key (first argument) already exists in a dictionary (second argument), if true, print the found item (key and value); if false, print "Not Found".

```python
def is_key_present(key, dic):
  if key in dic:
    print(key,dic[key])
  else:
    print("Not Found")
    
fruits = {'apple': 'red', 'pear': 'yellow', 'orange': 'orange', 'grape': 'purple'}
is_key_present('pear', fruits)
is_key_present('watermelon', fruits)
```

2. Create a function that takes a string as the input and returns a dictionary showing the number of characters in a string. <br>
Sample Input: 'apple pie' <br>
Output: {'a': 1, 'p': 3, 'l': 1, 'e': 2, ' ': 1, 'i': 1}

```python
def char_frequency(string):  
    d={}
    for ch in string:
        if ch in d:
            d[ch]=+1
        else:
            d[ch]=1
    return d

print(char_frequency('apple pie')) 
```

3. Create a function that takes a 'marks' input list, it remove the highest and lowest marks from the list and then return the average mark of the list.

```python
def cal_mean(marks):
    total=0
    max=marks[0]
    min=marks[0]

    for x in marks:
        total+=x
        if x>max:
            max=x
        elif x<min:
            min=x
    
    average=(total-max-min)/(len(marks)-2)
    return average
    
marks =[67, 66, 85, 23, 49, 72, 98, 57] 
print (cal_mean(marks))
    
```

4. Create a function that takes a string as the input and return a string that is the reverse of the input string. If the string length is less than 2, return 'The length of the input string is shorter than 2!'. <br>
Sample String : 'chau' <br>
Expected Result : 'uahc' <br>
Sample String : 'hi' <br>
Expected Result : 'ih' <br>
Sample String : '?' <br>
Expected Result : 'The length of the input string is shorter than 2!'

```python
def string_reverse(str):
    result=""
    if len(str)<2:
        print("The length of the input string is shorter than 2!")
        return ""
    for ch in str:
        result=ch+result
    return result
print(string_reverse('carnivore'))
print(string_reverse('hi'))
print(string_reverse('?'))
```

5. Create a function that takes two inputs of lists and return an interwined list in this way: <br>
Sample input: <br>
listA = [1, 2, 3, 4]<br>
listB = ['a', 'b', 'c', 'd', 'e', 'f']<br>
Return:
[1, 'a', 2, 'b', 3, 'c', 4, 'd', 'e', 'f'].<br>
Note: when the lengths do not match, the remainder shall be appended to the end of the output list. Also,you should use list comprehension in the function.

```python
def interwine(A, B):
    result=[]
    for a,b in zip(A,B):
        result.append(a)
        result.append(b)
    result.extend(A[len(B):])
    result.extend(B[len(A):])
    return result

        
listA = [1, 2, 3, 4]
listB = ['a', 'b', 'c', 'd', 'e', 'f']

print(interwine(listA, listB))
print(interwine(listB, listA))
print(interwine(listA, listA))
```

8. Sort a list of tuples by the first element in the tuple using Lambda.<br>
Sample input: [('Chiikawa', 10), ('Usagi', 5), ('Hachiware', 15)] <br>
Output: [('Chiikawa', 10), ('Hachiware', 15) , ('Usagi', 5)]

```python
def sort_tuples(before):
    result = sorted(before, key=lambda t: t[1], reverse=True)
    return result
before = [('Chiikawa', 10), ('Usagi', 5), ('Hachiware', 15)]
after = sort_tuples(before)
print(after)
```

9. Write a function to count the odd, even numbers in a given array of integers using Lambda. <br>
Sample input:  <br>
[1, 4, 3, 11, 7, 8, 15, 10]  <br>
Sample output:  <br>
Number of odd numbers: 5  <br>
Number of even numbers: 3  <br>

```python
def even_odd(array_numbers):
    even=0
    odd=0
    for x in array_numbers:
        if (lambda n:n%2==0)(x):
            even+=1
        else:
            odd+=1
    print(f"Number of odd numbers: {odd}\nNumber of even numbers: {even}")

    
array_example = [1, 4, 3, 11, 7, 8, 15, 10]
even_odd(array_example)
```

10. Complete the following program that calculates the area of different shapes using polymorphism.
    The program has a base class Shape.
    You need to create 3 subclasses (Rectangle, Triangle and Circle) which override the area() method with corresponding area calculations. Rectangle constructor should have the width and lenght parameters. Triangle constructor should have the base and height parameters. Circle constructor should have the radius parameter.

```python
import  math
class Shape:
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self,width,length):
        self.width=width
        self.length=length

    def area(self):
        return self.width*self.length

class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return math.pi * self.radius ** 2
    
class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return self.base * self.height / 2


# The following codes are for testing
shapes = [
    Rectangle(5, 10),
    Circle(7),
    Triangle(6, 8)]

for shape in shapes:
    print(f"{shape.__class__.__name__}: Area = {shape.area():.2f}")

        
```
