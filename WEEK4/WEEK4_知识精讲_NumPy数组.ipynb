## 1. 为什么使用 ndarray

Python 列表通用且方便；NumPy 数组面向数值数据，数组元素具有一致的数据类型，支持对数组进行逐元素运算。数组是可变的线性数据结构，可通过索引定位元素。

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([[1, 2, 3], [4, 5, 6]])
print(a, b)
```

## 2. 数组属性与创建

常见属性：`ndim` 是维度数，`shape` 是各轴长度，`size` 是元素总数，`dtype` 是元素数据类型。也可从头创建数组，例如全零、全一、指定范围或随机数组。NumPy 提供不同整数和浮点类型；具体使用哪种类型可查看 `dtype`。

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a.ndim, a.shape, a.size, a.dtype)
zeros = np.zeros((2, 3))
ones = np.ones((2, 2))
sequence = np.arange(0, 10, 2)
print(zeros, ones, sequence, sep="\n")
```

## 3. 索引和切片

一维索引与 Python 序列类似：起点包含、终点不包含，可带步长；负步长可反向读取。二维数组用 `[行, 列]` 指定位置，切片可同时选行和列。单独取一列会得到一维数组；用单元素切片保留该轴，可得到二维结果。

```python
x = np.arange(10)
print(x[2], x[:5], x[4:8], x[::2], x[::-1])

x2 = np.arange(1, 13).reshape(3, 4)
print(x2[0, -1])
print(x2[:2, :3])
print(x2[:, 0].shape)
print(x2[:, 0:1].shape)
```

布尔条件可以筛选或修改符合条件的元素。条件运算会得到与原数组对应的布尔数组，再将其用于索引。

```python
x2 = np.arange(1, 13).reshape(3, 4)
mask = x2 > 5
print(mask)
print(x2[mask])
x2[x2 > 10] = 0
print(x2)
```

## 4. 切片视图与复制

讲义特别提醒：从数组切片得到的子数组通常是原数组的视图。修改视图中的值会影响原数组。若需要彼此独立的数据，用 `copy()` 显式复制。

```python
parent = np.arange(1, 10).reshape(3, 3)
view = parent[:2, :2]
view[0, 0] = 100
print(parent)  # 原数组也变化

independent = parent[:2, :2].copy()
independent[0, 0] = -1
print(parent)
print(independent)
```

## 5. 变形、拼接与拆分

`reshape` 改变数组形状，元素总数必须匹配；加上 `None` 可在相应位置增加一个轴。拼接把多个数组组合起来，拆分把数组分成多个片段。

```python
x = np.array([1, 2, 3, 4, 5, 6])
print(x.reshape(2, 3))
print(x[None, :].shape)
print(x[:, None].shape)

left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])
print(np.concatenate([left, right], axis=0))
print(np.concatenate([left, right], axis=1))
print(np.split(x, 3))
```

## 6. 数组运算与统计

`+`、`-`、`*`、`/` 和幂运算按元素计算。点积把对应元素相乘后求和。转置交换数组轴的排列；二维数组常用 `.T`。常见统计操作包括最大值、最小值、均值、标准差，以及最大/最小值所在位置；指定轴可以逐行或逐列计算。

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b, a * b)
print(a.dot(b))

matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix.T, matrix.shape, matrix.T.shape)
print(matrix.max(), matrix.min(), matrix.mean(), matrix.std())
print(matrix.argmax(axis=1))
```

数组也可逐行迭代；讲义展示了 `enumerate` 为项目附上序号、`zip` 同步遍历多个序列。排序可用于一维或二维数组；二维数组可指定排序轴。需要留意原操作是返回排序结果还是直接改变数组。

```python
values = np.array([3, 1, 2])
print(np.sort(values))
print(list(enumerate(values)))
print(list(zip(values, values ** 2)))
```

### dtype 位宽说明与 resize

整数类型按位宽表达不同范围：int8 为 -128–127，int16 为 -32768–32767，int32/int64 对应更宽范围；uint8/16/32/64 是非负无符号整数。float16/32/64 分别是半精度、单精度、双精度浮点；讲义列出 float16 的 5 位指数/10 位尾数，float32 的 8/23 位，float64 的 11/52 位。complex64 由两个 float32 表示，complex128 由两个 float64 表示；int_、intc、intp 分别对应默认整数、C int、索引整数类型。resize 示例将数组形状改为目标尺寸；讲义展示扩展形状时用 refcheck=False，并以 0 补新增元素。

精确范围：int8 -128–127，int16 -32768–32767，int32 -2147483648–2147483647，int64 -9223372036854775808–9223372036854775807；uint8 0–255，uint16 0–65535，uint32 0–4294967295，uint64 0–18446744073709551615。int_ 是默认整数（通常 int64 或 int32），intc 对应 C int，intp 用作索引；float_ 是 float64 的简写，complex_ 是 complex128 的简写。
