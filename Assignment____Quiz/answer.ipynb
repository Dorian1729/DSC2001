```python
import pandas as pd
import numpy as np
import math

L1 = list([3*i for i in range (100) ])
L2 = list([i*i for i in range (100) ])
L3 = list([math.cos(i) for i in range (100) ])
L4 = list([math.sqrt(i) for i in range (100) ])
s1 = pd.Series(L1,dtype='float64')
s2 = pd.Series(L2,dtype='float64')
s3 = pd.Series(L3,dtype='float64')
s4 = pd.Series(L4,dtype='float64')
```

### Task 05:
1. Add s1 and s2 and name the resulted series as s5
2. Multiply s3 by 100 and name the resulted series as s6
3. Take the exponetial of s4 and name the resulted series as s7

```python
s5 = s1 + s2
s6 = s3 * 100
s7 = np.exp(s4)
```

### Task 06:
1. Create a DataFrame df1 by including the two Series s3, s4 as columns, and the column names should be 'Cos' and 'Sqrt'. Then, **drop the first 50 rows** of df1 (shape is (50,2)).
3. Create another DataFrame df2 by including the three Series s5, s6 and s7 as three columns, and name the columns as 'a', 'b' and 'c' respectively. **Keep the first 70 rows** of df2 (shape is (70,2)).
4. Using the concat() method, join (outer) df1 and df2 in vertical direction to form df3 (shape is (120,5)) with clean (no duplicate) row index. Replace all empty cells with 0.
5. Merge df1 and df2 by the row index into df4 (shape is (20,5)). We only want to keep the intersections of the two dataframes.
6. Unstack df4 to an multiIndex series s8 (length is 100).

```python
df1 = pd.DataFrame({'Cos': s3, 'Sqrt': s4}).iloc[50:].copy()
df2 = pd.DataFrame({'a': s5, 'b': s6, 'c': s7}).iloc[:70].copy()
df3 = pd.concat([df1, df2], axis=0, ignore_index=True).fillna(0)
df4 = df1.join(df2, how='inner')
s8 = df4.unstack()
```

### Task 08: 
1. Execute the above instructions to read in the attached data.csv file. Study the features.<br>
2. Create a new sales revenue column:  revenue = quantity * unit_price * (1 - discount).
3. Create pivot table ( name it as 'pivot_qty') to show the 'quantity' sums of different salespersons and categories.
4. Create a 'salesperson' group object , name it as 'salesperson_group'.
5. Add a column 'region_avg_revenue' containing the average revenue of that region, on every row belonged to that region.<br>  Hint: groupby(...).transform('mean').

```python
df['revenue'] = df['quantity'] * df['unit_price'] * (1 - df['discount'])
pivot_qty = pd.pivot_table(
    df, values='quantity', index='salesperson', columns='category',
    aggfunc='sum', fill_value=0
)
salesperson_group = df.groupby('salesperson')
df['region_avg_revenue'] = df.groupby('region')['revenue'].transform('mean')
```
