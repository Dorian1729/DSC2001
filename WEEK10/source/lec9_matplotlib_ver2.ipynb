### Simple line plot

## Implicit (Pyplot) approach :
- Matplotlib implicitly creates a figure and an axes for you if they don't already exist.
- You can create multiple figures and axes
- Matplotlib's pyplot module maintains global state: at any moment there is one current figure (plt.gcf()) and, within it, one current axes (plt.gca()).
- Tracks the "current" figure and axes automatically behind the scenes.
- Most plt.* functions act on these "current" figure and axes implicitly.
- Use plt.figure(num) to create or switch between figures
- Use plt.sca(ax)  to  switch between axes


```python
plt.plot(x, np.sin(x))  # plot on "current figure" and "current axes" 
```

```python
#call it multiple times for multiple lines
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x));
```

## Explicit (Object-Oriented) approach:
* explicitly create and reference ploting objects (fig and ax). <br>
* code exposes fig (canvas) and ax (individual plot area).
* this is cleaner, more predictable, and essential for complex plots with multiple subplots.

```python
#create figure and axes
fig = plt.figure()
ax = plt.axes()
x = np.linspace(0, 10, 1000)
#use the ax.plot after creating axes
ax.plot(x, np.sin(x));
```

```python
plt.figure(figsize=(10, 4))  # set the cuurent canvas width in 10 inches and height in 4 inches

# Create the first axes (left side)
ax1 = plt.axes([0.1, 0.1, 0.4, 0.8]) # [left, bottom, width, height]
'''
left: 0.1 - 10% from the left edge of the figure
bottom: 0.1 - 10% from the bottom edge of the figure
width: 0.4 - 40% of the figure width
height: 0.8 - 80% of the figure height
'''
ax1.plot([1, 2, 3], 'r-') # Red line
ax1.set_title('Left Plot')

# Create the second axes (right side)
ax2 = plt.axes([0.55, 0.1, 0.4, 0.8])
ax2.bar(['A', 'B', 'C'], [3, 7, 2]) # Bar chart
ax2.set_title('Right Plot')
```

#### Colors and styles

```python
plt.plot(x, np.sin(x - 0), color='blue') # specify color by name
plt.plot(x, np.sin(x - 1), color='g') # short color code (rgbcmyk)
plt.plot(x, np.sin(x - 2), color='0.75') # Grayscale between 0 and 1
plt.plot(x, np.sin(x - 3), color='#FFDD44') # Hex code (RRGGBB from 00 to FF)
plt.plot(x, np.sin(x - 4), color=(1.0,0.2,0.3)) # RGB tuple, values 0 and 1
plt.plot(x, np.sin(x - 5), color='chartreuse'); # all HTML color names supported
plt.show()
```

1. Common named colors include:
'red', 'blue', 'green', 'black', 'cyan', 'magenta', 'yellow', 'white', 'orange', 'purple', 'gray', 'brown'.
2. Single-character shorthand (basic colors only):
'b' for blue, 'g' for green, 'r' for red, 'c' for cyan, 'm' for magenta, 'y' for yellow, 'k' for black, 'w' for white
3. Hexadecimal Color Codes: '#RRGGBB' where RR, GG, and BB are hexadecimal values (00-FF).
4. RGB Tuples: Values between 0 and 1
5. Grayscale Values:  a string to specify a shade of gray; a float between 0 (black) and 1 (white)


```python
plt.plot(x, x + 0, linestyle='solid')
plt.plot(x, x + 1, linestyle='dashed')
plt.plot(x, x + 2, linestyle='dashdot')
plt.plot(x, x + 3, linestyle='dotted');
# For short, you can use the following codes:
plt.plot(x, x + 4, linestyle='-') # solid
plt.plot(x, x + 5, linestyle='--') # dashed
plt.plot(x, x + 6, linestyle='-.') # dashdot
plt.plot(x, x + 7, linestyle=':'); # dotted
plt.show()
```

```python
##combined
plt.plot(x, x + 0, '-g') # fmt string: [marker][line][color] = '-g'; solid green ; 
plt.plot(x, x + 1, '--c') # dashed cyan
plt.plot(x, x + 2, '-.k') # dashdot black
plt.plot(x, x + 3, ':r'); # dotted red
plt.show()
```

### Errorbars

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 50)
dy = 0.8
y = np.sin(x) + dy * np.random.randn(50)
plt.errorbar(x, y, yerr=abs(np.random.randn(50)), fmt='.k')  # yerr must be positive ; so use abs()
plt.show()

#fmt: format code controlling the appearance of lines and points as in plt.plot
```

```python
#finer control
plt.errorbar(x, y, yerr=dy, fmt='ok',  
             ecolor='lightgray', elinewidth=2,     # error line
             capsize=4)      #length of the error bar caps in points
plt.show()
```

#### Two-dimensional histogram

```python
# if two variables have some correlation
mean = [0, 0]
cov = [[1, 1], [1, 2]] 
z = rng.multivariate_normal(mean, cov, 10000) #create 2 groups of normaly distributed random numbers with mean [0,0] ; 
# the variances of each group and the covarianc between each pair of elements of the two groups are specified by the covariance matrix 
# z is a 10000x2 array
x, y = z.T  # Transpose the a 2x10000 array, then unpack the resulting array to become 2 1D arrays
plt.hist2d(x, y, bins=30, cmap='Blues')
cb = plt.colorbar(label='counts in bin')
plt.show()
```

### Kernel Density Estimation

```python
# 1-D Data
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# Sample data
rng = np.random.default_rng(1)
data = rng.normal(0, 1, size=1000)

# Using scipy
kde = gaussian_kde(data) # estimates the continuous probability distribution function (kde) that generated 'data'
x = np.linspace(-5, 5, 1000)
plt.plot(x, kde(x), label='KDE (Scipy)')
plt.hist(data, bins=30, density=True, alpha=0.3, label='Histogram')
plt.legend()
plt.show()
```

```python
# 2-D Data Using Seaborn library
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
mean = [0, 0]
cov = [[1, 1], [1, 2]] 
z = rng.multivariate_normal(mean, cov, 10000) #create 2 groups of normaly distributed random numbers with mean [0,0] ; 
# the variances of each group and the covarianc between each pair of elements of the two groups are specified by the covariance matrix 
# z is a 10000x2 array
x, y = z.T  # Transpose the array and  than unpack the 2x10000 resulting array to become 2 1D arrays

sns.kdeplot(x=x, y=y, fill=True, cmap="viridis", levels=10)  #the number and placement of contour lines = 10
plt.xlabel("X-axis Label")
plt.ylabel("Y-axis Label")
plt.title("2D Kernel Density Estimate")
```

#### Subplot

```python
# Plot a sine and cosine curve on the same axe
fig, ax = plt.subplots()
x = np.linspace(0, 3 * np.pi, 1000)
ax.plot(x, np.sin(x), lw=1, label='Sine')
ax.plot(x, np.cos(x), lw=3, label='Cosine')

# Set up grid, legend, and limits
ax.grid(True)
ax.legend(frameon=False)
ax.axis('equal')
ax.set_xlim(0, 3 * np.pi);

plt.show()
```

```python
# create  multiple plots in a single figure and arrange plots in a grid layout 
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))  
ax_left, ax_right = axes  # axes is a 1D numpy array: [ax_left, ax_right]
x = np.arange(1,10)  
y1= x    # linear_data
y2= x**2   # exponential_data = linear_data**2 

ax_left.plot(x, y1, color="tab:blue", linewidth=2, linestyle="--", label="linear")
ax_left.legend()
ax_right.plot(x, y2, color="tab:red", linewidth=2, linestyle="--", label="exponential)")
ax_right.legend()

plt.show()
```

#### Text and Annotation

```python
fig, ax = plt.subplots(2, 3, sharex='col', sharey='row')
# All x-axes in the same column have identical limits, and only the bottom row shows x tick labels
# All y-axes in the same row have identical limits, and only the first column shows y tick labels
# ax is an array of shape (2,3) and accessed by ax[i,j] array indexes
for i in range(2):
    for j in range(3):
        ax[i, j].text(0.5, 0.5, #location
                      str((i, j)), #text
                     fontsize=18, 
                     ha='center', #horizontal alignment
                     color='red')
plt.show()      
```
