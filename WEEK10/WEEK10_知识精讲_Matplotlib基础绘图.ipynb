## 1. Figure 与 Axes

Figure 是整张图的画布，Axes 是其中一个绘图区域。Pyplot 隐式方式由 `plt` 管理当前画布和坐标轴，适合简单图形；面向对象方式显式创建并使用 `fig` 与 `ax`，多个绘图区时更清晰。

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [2, 5, 4, 8]
fig, ax = plt.subplots()
ax.plot(x, y, marker="o")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("Simple line plot")
plt.show()
```

## 2. 折线图与散点图

`plot(x, y)` 根据 x、y 坐标画点，并默认用线连接。格式字符串可组合颜色、标记和线型；也可分别用参数设定。折线图适合连接数据点；`scatter` 用于单独控制每个点的位置、大小或颜色。

```python
fig, ax = plt.subplots()
ax.plot([1, 2, 3], [3, 2, 5], "ro--", label="Series A")
ax.scatter([1, 2, 3], [2, 4, 3], color="blue", s=35, label="Series B")
ax.set_xlim(0, 4)
ax.set_ylim(0, 6)
ax.set_xlabel("Observation")
ax.set_ylabel("Value")
ax.legend()
plt.show()
```

颜色可写常用英文名称、单字符缩写或十六进制颜色码。图例由绘图项目的 label 与 `legend()` 显示；坐标轴范围、标题和轴标签帮助读者理解图中变量。

## 3. 误差线与分布图

误差线可与点或柱关联，用于显示连续数据的误差范围。直方图把数值范围切成 bins 并显示每个区间的频数；二维直方图同时按两个变量分箱。KDE 用平滑曲线展示分布形状。

```python
import numpy as np

rng = np.random.default_rng(1)
scores = rng.uniform(0, 100, 200)
fig, ax = plt.subplots()
ax.hist(scores, bins=10, color="lightblue", edgecolor="black")
ax.set_xlabel("Score")
ax.set_ylabel("Count")
plt.show()
```

绘图时通过 `errorbar` 添加误差线，通过 `hist2d` 或相关绘图接口查看二维分布；课程材料也示范在 Seaborn 中绘制 KDE。

```python
x = np.arange(1, 6)
means = [22, 30, 35, 35, 26]
errors = [4, 3, 4, 1, 5]
fig, ax = plt.subplots()
ax.errorbar(x, means, yerr=errors, fmt="o")
plt.show()
```

## 4. 子图、注释与保存

子图把多个 Axes 排列在同一 Figure 中，便于并排比较。文字注释可直接放入绘图区。用 Figure 的 `savefig` 保存图像；可查看当前绘图格式支持列表。

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].plot([1, 2, 3], [2, 3, 5])
axes[0].annotate("point", xy=(3, 5))
axes[1].hist(scores, bins=10)
fig.tight_layout()
fig.savefig("my_figure.png")
plt.show()
```

### 补充：Notebook 显示、KDE、二维直方图与图例

source 说明 `%matplotlib inline` 可将静态图直接嵌入 Notebook。除一维直方图外，还包括二维直方图和 KDE；图例可进一步调整位置与显示方式，子图中的文字和注释用于指出特定位置或内容。`savefig` 将图保存为静态图像。

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].hist2d([1, 1, 2, 3], [1, 2, 2, 3], bins=4)
axes[1].plot([1, 2, 3], [2, 4, 3], label="scores")
axes[1].legend(loc="best", title="Series")
axes[1].annotate("peak", xy=(2, 4))
plt.show()
```
