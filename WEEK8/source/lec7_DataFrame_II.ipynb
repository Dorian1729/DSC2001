## MultiIndex
A multi-index (also known as a hierarchical index) in pandas allows you to have multiple index levels on an axis, enabling higher-dimensional data representation in a 2D  structure.

```python
import pandas as pd
import numpy as np
# using index tuples rather than multi-level index
index_tuples = [('California', '2000'), ('California', '2010'),
            ('New York', '2000'), ('New York', '2010'),
            ('Texas', '2000'), ('Texas', '2010')]
data = {'populations' : [33871648, 37253956,
                    18976457, 19378102,
                    20851820, 25145561],
        'income' :[ 52396,  69562, 
                61297, 70312, 
                49995, 55340]}

pop = pd.DataFrame(data, index=index_tuples)
pop
```

```python
pop.loc['California']  # Error
```

```python
pop.index
```

```python
pop.loc[[('California','2010')]] 
```

```python
# for multiple rows selection, need to use list comprehension to create the index list
pop.loc[[i for i in pop.index if i[0]=='California']]
```

```python
# better way, change the index to multi_index
multi_index = pd.MultiIndex.from_tuples(index_tuples, names=['State', 'Year'])
pop = pop.reindex(multi_index)
pop
```

```python
pop.index   # index is changed to multi-level index
```

```python
# we can still use index list to select records that meet selection criteria
pop.loc[[i for i in pop.index if i[0]=='California']]
```

```python
# select group by just the first level index  
pop.loc['California']
```

```python
#selection using Boolean Indexing (Filtering)
mask = pop.income > 60000
pop.loc[mask]
```

```python
# select any row by index tuple
pop.loc[('California','2010')] 
```

```python
# selection based on the second index, i.e. 2010 is not allowed
pop.loc['2010']   # Error
```

```python
# Using xs() method (Cross-Section) for selecting data at a particular level
pop.xs('2010', level='Year')
```

```python
pop.columns
```

```python
pop.stack()   # add one more index level; flat => tall 
```

```python
# "moves" the innermost row index back into the column labels, making the DataFrame wider.
pop_df = pop.unstack()  
pop_df                      # row index : 'State';   column index tuples: (None, 'Year')
```

```python
pop_df.unstack()   # become  a mutiIndex Series
```

### stacking and unstacking
**Stacking** "pivots" from column level to row level, converting a DataFrame with wide format to a long format
; converts a Dataframe to a Multi-indexed Series. <br>
**Unstacking** "pivots" from row level to column level, converting a Series with MultiIndex back to a DataFrame.

```python
df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6],
    'C': [7, 8, 9]
}, index=['X', 'Y', 'Z'])

print("Original DataFrame:")
print(df)
```

```python
stacked = df.stack()  # Flat => Tall ; column names becone 2nd level (level 1) indexes
stacked
```

```python
stacked.unstack()  # Tall return back to Flat ; inner row names becone column indexes
```

```python
df.unstack()   #  row names becone inner column indexes (multIndex Series)
```

```python
pop
```

```python
pop.unstack(level=0) #level 0 row indexes converted to level 1 column indexes 
```

```python
pop.unstack(level=1) #level 1 row indexes converted to level 1 column names 
```

```python
pop.unstack()  # default is level 1
```

```python
pop.unstack(level=1).stack()   # unchanged
```

```python
pop.unstack(level=0).stack()  # 1st level indexes <=> 2nd level indexes
```

#### Data Aggregation

```python
import pandas as pd
data = {
    'Region': ['North', 'North', 'South', 'South', 'North', 'South'],
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250, 300, 350],
    'Profit': [20, 30, 40, 50, 60, 70]
}
df = pd.DataFrame(data)
print(df)
# Group by multiple columns 
multi_group_obj = df.groupby(['Region', 'Category'])  #the resulting object's items can't be accessed directly
    
multi_group_sum = multi_group_obj.sum()

 # similar to SQL: select  sum('Sales'), sume('Profit') from df group by ('Region', 'Category')
print(multi_group_sum)
```

```python
# Note: 'Region', 'Category' become multilevel indexes; 
# if you don't want them to be indexes; set the parameter as_index= False:
df.groupby(['Region', 'Category'], as_index= False).sum()
```

```python
# Multiple aggregations

result = df.groupby(['Region', 'Category']).agg({
    'Sales': ['mean', 'sum', 'std'],
    'Profit': ['min', 'max', 'count']
})
print(result)
```

```python
result = df.groupby( 'Category').agg({
    'Sales': ['mean', 'sum', 'std'],
    'Profit': ['min', 'max', 'count']
})
print(result)
```

```python
#Note: groupby() create a DataFrameGroupBy object (iterator), we can print the values by for loop
gp= df.groupby( 'Category')
gp

for name,group in gp:
    print(name)
    print (group)
    print()
```
