## 1. NumPy 伪随机数

随机数生成器产生的是伪随机数。讲义展示用 NumPy 随机功能抽样、指定输出形状，以及设置离散结果各自出现的概率。生成的数值可直接作为 NumPy 数组用于后续计算或绘图。

```python
import numpy as np

rng = np.random.default_rng(1)
print(rng.random(5))
print(rng.integers(0, 10, size=(3, 4)))
print(rng.choice([0, 1, 2], size=10, p=[0.2, 0.5, 0.3]))
```

## 2. 常见分布

讲义列出若干用于模拟不同结果模式的分布。每个分布的参数说明抽样机制和结果形状；这些例子用于生成随机数据，不是对真实数据的结论。

### 正态分布

正态分布由中心位置 `loc` 和标准差 `scale` 描述；`size` 决定返回数组的形状。

```python
normal = rng.normal(loc=0, scale=1, size=1000)
```

### 二项分布与多项分布

二项分布描述重复进行二元试验时某类结果出现的次数：`n` 为试验次数，`p` 为单次发生概率。多项分布扩展到多个可能结果，`pvals` 给出各结果概率。

```python
coin_counts = rng.binomial(n=10, p=0.5, size=8)
dice_counts = rng.multinomial(n=12, pvals=[1/6] * 6, size=3)
print(coin_counts, dice_counts, sep="\n")
```

### 泊松分布与指数分布

泊松分布用于描述给定时间或范围内事件出现的次数，参数 `lam` 表示平均发生率。指数分布用于描述事件之间的时间间隔；讲义将其与泊松过程联系，并用 `scale` 表示间隔尺度。

```python
events = rng.poisson(lam=3, size=10)
intervals = rng.exponential(scale=1/3, size=10)
print(events)
print(intervals)
```

### Logistic 与 Uniform 分布

Logistic 分布由 `loc` 和 `scale` 控制位置与形状，讲义指出其尾部比正态分布更重。Uniform 分布表示指定区间内各值具有相同抽样机会，使用 `low` 和 `high` 确定范围。

```python
logistic = rng.logistic(loc=0, scale=1, size=1000)
uniform = rng.uniform(low=0, high=1, size=1000)
```

## 3. 查看随机分布

source 使用 Seaborn 绘图查看随机数据的分布。直方图展示抽样值的频数形状；将分布参数和样本放在一起观察，可以帮助理解抽样结果。

```python
import matplotlib.pyplot as plt
import seaborn as sns

sample = rng.normal(loc=0, scale=1, size=1000)
sns.histplot(sample, kde=True)
plt.show()
```

## 4. NumPy ufunc

ufunc 是可作用于 ndarray 元素的通用函数，便于以数组方式执行运算。讲义说明 NumPy 运算会向量化，不必逐项手写循环；也可通过 `frompyfunc()` 将自定义 Python 函数包装成 ufunc。

```python
def add_one(value):
    return value + 1

add_one_ufunc = np.frompyfunc(add_one, 1, 1)
data = np.array([1, 2, 3])
print(add_one_ufunc(data))
```

### 随机数生成器与分布可视化补充

source 展示两种 NumPy 随机数接口：旧式 np.random（例如先 seed(1) 再调用 rand、randint），以及 default_rng(seed) 返回的生成器。可生成一维、二维和三维数组；permutation 返回随机排列，shuffle 原地打乱输入，choice 可抽样并指定概率 p。固定 seed 用于复现实验中的同一组随机结果。

Seaborn 分布图示例包含 histogram、KDE 和 ECDF（经验累积分布函数）。正态、二项、泊松、指数、Logistic、Uniform、多项分布都可用样本数组生成并绘图。

### ufunc 的逐元素行为

NumPy 内置 ufunc（如 np.add）对数组逐元素运算，可替代对多个序列逐项循环。np.frompyfunc(function, nin, nout) 将自定义函数包装成 ufunc；nin 是输入数组数量，nout 是输出数组数量。
