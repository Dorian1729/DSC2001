## 2. Python 数据集合

集合类型用于在一个变量中保存多个项目。选类型时先考虑项目是否有序、是否需要修改、是否允许重复，以及如何访问项目。

| 类型 | 特点 | 写法 |
|---|---|---|
| List 列表 | 有序、可修改、允许重复 | `[...]` |
| Tuple 元组 | 有序、不可修改、允许重复 | `(...)` |
| Set 集合 | 不按位置索引、元素不重复；可增删 | `{...}` |
| Dictionary 字典 | 键值对、有序、可修改；键不重复 | `{key: value}` |

### List 列表

列表索引从 0 开始，负索引从末尾往前数，`-1` 是最后一项。切片 `[start:stop]` 包含起点、不包含终点。列表可修改，也能混合不同类型。切片赋值会替换该段内容，所以可以用两个新项目替换一个旧项目。

```python
items = ["abc", 34, True, 40, "male"]
print(items[0], items[-2])
print(items[2:4])
items[1] = 36
items[2:3] = [False, "Hong Kong"]
print(items)
```

追加、插入、扩展分别用于在末尾加入一个项目、在指定位置插入、加入另一个可迭代对象中的项目。移除可按值或索引操作。Tutorial 还列出排序、清空、删除，以及用 `copy()` 或切片复制列表。

```python
fruits = ["apple", "cherry", "banana"]
fruits.append("kiwi")
fruits.insert(1, "orange")
fruits.extend(["mango", "pear"])
fruits.remove("cherry")
removed = fruits.pop(0)
last = fruits.pop()
print(fruits, removed, last)
```

列表推导式把“对每个项目计算结果”写成一行，可加筛选条件。一般形式是 `[expression for item in iterable if condition]`；条件可省略。

```python
squares = [x * x for x in range(1, 6)]
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
with_a = [x for x in fruits if "a" in x]
print(squares, with_a)

# 同样的筛选逻辑也可写成循环
with_a_loop = []
for x in fruits:
    if "a" in x:
        with_a_loop.append(x)
```

### Tuple 元组

元组可像列表一样按索引读取，但创建后不能直接修改项目或长度。若要修改，Tutorial 示范先转成列表，修改后再转回元组；元组也可用 `+` 合并。

```python
values = ("apple", "banana", "cherry")
print(values[-1])
temporary = list(values)
temporary[-1] = "kiwi"
values = tuple(temporary)
combined = values + ("pear",)
print(combined)
```

### Set 集合

集合不支持按索引访问，元素不重复。可用 `in` 检查成员，或用循环遍历；可添加、更新、移除和清空。并集收集两边所有项目，交集保留共同项目，差集保留左侧独有项目，对称差集保留只出现在一侧的项目。

```python
x = {"apple", "banana", "cherry"}
y = {"apple", "oracle"}
x.add("mango")
x.update(["kiwi", "orange"])
x.discard("banana")
print("apple" in x)
print(x | y)  # 并集
print(x & y)  # 交集
print(x - y)  # 差集
print(x ^ y)  # 对称差集
```

### Dictionary 字典

字典用键值对保存数据。通过键读取值；给不存在的键赋值会新增项目，已有键则更新。讲义说明字典按插入顺序保存项目。常见操作包括查看键、复制、按键移除并返回值，以及清空。

```python
person = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(person["brand"])
person["color"] = "red"
person.update({"year": 2020})
keys_view = person.keys()
copy_of_person = person.copy()
removed = person.pop("color")
print(keys_view, removed, copy_of_person)
```

## 3. 函数

函数将一组操作命名，便于重复调用。定义中的参数接收调用时传入的实参；`return` 将结果交回调用处。OOP 讲义用平方和说明：把重复计算写成函数，通过参数传入不同数据，并把每次结果返回给调用者。

```python
def sum_squares(a, b):
    return a * a + b * b

first = sum_squares(2, 3)
second = sum_squares(4, 5)
print(first, second)
```

参数可按位置或名称传入，也可设置默认值。Tutorial 还介绍 `*names` 收集任意数量的位置参数（函数内为元组），`**name` 收集任意数量的关键字参数（函数内为字典）。参数列表中的 `/` 表示其左侧参数只能按位置传入；`*` 表示其右侧参数只能按名称传入；中间的参数可以按位置或名称传入。

```python
def describe(*names):
    for name in names:
        print(name)

def show_name(**name):
    print(name["lname"])

describe("Emil", "Tobias", "Linus")
show_name(fname="Tobias", lname="Refsnes")

def calculate(a, b, /, standard, *, c, d=9):
    print(standard)
    return a + d - b - c

print(calculate(5, 6, "S", c=7))
```

## 4. 类与对象

类把相关的数据和操作放在一起，对象是按类创建的实例。`__init__` 在创建实例时设置属性；`self` 指向当前实例，用于访问该对象的属性和方法；`__str__` 定义打印对象时显示的文字。

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}({self.age})"

    def introduce(self):
        print("Hello my name is " + self.name)

p1 = Person("John", 36)
print(p1)
print(f"{p1.name} with age {p1.age}")
p1.introduce()
```

### 面向对象的组织方式

讲义用幂次和的例子说明面向对象的动机：若平方和、立方和等分别写成函数，支持更多幂次就会不断增加函数。把幂次作为对象属性、把计算放进对象方法后，可以创建不同幂次的对象并调用相同的 `calculate` 操作。

```python
class SumPower:
    def __init__(self, power_order):
        self.power_order = power_order

    def power(self, value):
        return value ** self.power_order

    def calculate(self, x, y):
        return self.power(x) + self.power(y)

sum_squares = SumPower(2)
sum_cubes = SumPower(3)
print(sum_squares.calculate(2, 3))
print(sum_cubes.calculate(2, 3))
```

### 封装、继承、多态与抽象

- **封装**：把属性和方法放在类中，并控制外部如何使用。讲义建议隐藏内部数据，通过公开方法（API）提供必要操作；需要时可提供 get/set 方法。
- **继承**：子类沿用父类属性与方法，也可添加功能。Tutorial 用空的 `Student(Person)` 展示继承；讲义用 `super()` 调用父类构造方法。
- **多态**：不同对象可响应同一方法名而表现不同；子类可以覆盖父类方法。
- **抽象**：呈现必要的功能入口，把具体实现细节留在内部或交给子类。抽象类可有已实现的方法，也可声明待实现的抽象方法。

讲义用汽车比喻区分封装和抽象：封装关注内部数据如何受到保护；抽象关注使用者看到的简明操作接口。

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Move!")

class Car(Vehicle):
    pass

class Boat(Vehicle):
    def move(self):
        print("Sail!")

for vehicle in (Car("Ford", "Mustang"), Boat("Ibiza", "Touring 20")):
    print(vehicle.brand, vehicle.model)
    vehicle.move()
```

OOP 讲义用 `ABC` 和 `@abstractmethod` 展示 Python 中的抽象类：抽象方法声明子类要提供的操作，子类实现后再创建实例。讲义也比较抽象类与接口：抽象类可包含已实现方法，接口表达一组能力约定；Python 没有 Java 式的 `interface` 关键字，材料用 ABC 展示相近表达。

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def drive(self):
        pass

class Car(Vehicle):
    def drive(self):
        print("Driving")

car = Car()
car.drive()
```

## 6. 异常处理

异常是程序运行中出现的错误。Tutorial 介绍四部分：`try` 放可能出错的语句；`except` 在异常发生时处理它；`else` 在没有异常时执行；`finally` 无论是否发生异常都会执行。

```python
try:
    value = int("12")
except ValueError:
    print("输入无法转换为整数")
else:
    print("转换成功：", value)
finally:
    print("处理结束")
```

可以用 `raise` 主动引发异常；也可以按异常类型设置不同处理分支。

```python
x = -1
if x < 0:
    raise Exception("Sorry, no numbers below zero")
```

### 补充：集合操作、多态与异常
