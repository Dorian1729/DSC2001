**None** is a special data type (NoneType) to represent missing/null values . <br>
**NaN** , not a number, is a specific floating point value used to represent any numerical value that is undefined or unpresentable. <br>
In any mathmatical computation, NaN is returned if the calculation has any NaN argument instead of TypeError (Safe Execution). <br>
One has to be mindful that in Python (and NumPy), the NaNs don’t compare equal, but None do. 

```python
np.nan == None  # they are two different things
```

```python
np.nan == np.nan #can't be compared using '=='
```

```python
a = np.nan
print (np.isnan(a))    # use isna() to check if a variable has NaN value
```

```python
a = None
a == None    # 'None' variable can be checked using '=='
```

### Think of a Pandas Series like a specialized Python dictionary:

A dictionary is a structure that maps arbitrary keys to a set of arbitrary values; <br>
a Series is a structure that maps typed keys to a set of typed values. 

**typed**: just as the type-specific compiled code behind a NumPy array makes it more efficient than a Python list for certain operations, the type information of a Pandas Series makes it much more efficient than Python dictionaries for certain operations.

```python
# constructing a Series object directly from a Python dictionary
population_dict = {'California': 38332521,
                    'Texas': 26448193,
                    'New York': 19651127,
                    'Florida': 19552860,
                    'Illinois': 12882135}
population = pd.Series(population_dict)  # dictionary should have typed keys and values
population
```

```python
#the index is drawn from the keys
population.index
```

```python
#Unlike a dictionary, the Series also supports 
#array-style operations such as slicing
population['California':'Florida']
```

Be careful if your Series has an explicit integer index: <br>
An indexing operation such as data[1] will use the explicit indices; <br>
a slicing operation like data[1:3] will use the implicit Python-style index.

```python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
```

```python
# explicit index when indexing
data[1]
```

```python
# implicit index when slicing
data[1:3] #data[1:3] (implicit index)
```

**To avoid confusion, use special indexer:** <br>
**loc[explicit index] ; iloc[implicit index]**

```python
#loc attribute always references the explicit
print(data.loc[1],'\n')
print(data.loc[1:3])
```

```python
#iloc attribute always references the implicit
print(data.iloc[1],'\n')
print(data.iloc[1:3])
```
