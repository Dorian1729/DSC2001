### The DataFrame Data Structure

If a Series is an analog of a one-dimensional array with flexible indices, a DataFrame is an analog of a two-dimensional array with both flexible row indices and flexible column names. 

A two-dimensional array can be viewed as an ordered sequence of aligned one-dimensional columns; a DataFrame can be viewed as a
sequence of aligned Series objects sharing the same index.

```python
import pandas as pd
area_dict = {'California': 423967, 
             'Texas': 695662, 
             'New York': 141297,
             'Florida': 170312, 
             'Illinois': 149995}
area_series = pd.Series(area_dict)

population_dict = {'California': 38332521,
                    'New York': 19651127,
                    'Texas': 26448193,
                    'Florida': 19552860,
                    'Illinois': 12882135}
population_series = pd.Series(population_dict)

states = pd.DataFrame({'population': population_series,
'area': area_series})
states
```

```python
states.index
```

```python
# additional columns attribute
states.columns
```

```python
#a pandas dataframe column is a Series
states['area']
```

```python
states['Texas'] #Error! it is a column indexing format, not row indexing format
```

```python
#a pandas dataframe row  is a Series
states.loc['Texas']
```

```python
##notice the difference for indexing between numpy array and pandas dataframe
import numpy as np
a = np.random.randint(0, 10, (2,3))
print(a)
print(a[0])   # row indexing
print(a[:,0])  #column indexing
```

### Series Index alignment
The resulting Series contains the union of indices of the two input Serieses; <br>
Any item for which one or the other does not have an entry is marked with NaN, or "Not a Number".

```python
import pandas as pd
area = pd.Series({'Alaska': 1723337, 'Texas': 695662,
                    'California': 423967}, name='area')
population = pd.Series({'California': 38332521, 'Texas': 26448193,
                    'New York': 19651127}, name='population')
population / area
```

```python
area.index.union(population.index) #combine the index of the two series
```

```python
area.index.intersection(population.index) #match the index of the two series
```

```python
# another example:
X = pd.Series([2, 4, 6], index=[0, 1, 2])
Y = pd.Series([1, 3, 5], index=[1, 2, 3])
X + Y
```

If using NaN values is not the desired behavior, we can modify the fill value using appropriate object methods in place of the operators. For example, calling X.add(Y) is equivalent to calling X + Y, but allows optional explicit specification of the fill value for any elements in X or Y that might be missing:

```python
X.add(Y, fill_value=0)
```

### Pandas: Missing data

NaN , not a number, is a numeric data type used to represent any value that is undefined or unpresentable. <br>
It is a special floating-point value recognized by all systems that use the standard IEEE floating-point representation. <br>
NaN is also assigned to variables, in a computation, that do not have values and have yet to be computed.

```python
numbers = np.array([1, 2, None])
print(pd.Series(numbers))
```

```python
#none returns error
numbers.sum()
```

```python
numbers = np.array([1, 2, np.nan])
print(pd.Series(numbers))
```

```python
#nan doesn't return errors, but not useful
numbers.sum()
```

```python
1 + np.nan
```

```python
0 * np.nan
```

```python
#use special aggregations; ignore the NaNs
np.nansum(numbers), np.nanmin(numbers), np.nanmax(numbers)
```

One has to be mindful that in Python (and NumPy), the nan's don’t compare equal, but None's do. 

```python
import numpy as np
np.nan == None
```

```python
None == None
```

```python
np.nan == np.nan
```

```python
np.isnan(np.nan)
```

### NaN in Pandas
For types that don’t have an available sentinel value, Pandas automatically type-casts when NA values are present. <br>
If we set a value in an integer array to np.nan, it will automatically be upcast to a floating-point type to accommodate the NA. <br>

```python
pd.Series([1, np.nan, 2, None])
```

```python
x = pd.Series(range(2), dtype=int)
print(x)
```

```python
x[0] = None
x
```

**Pandas handing None by type** <br>
floating **No change** np.nan <br>
object **No change** None or np.nan <br>
integer **Cast to float64** np.nan <br>
boolean **Cast to object** None or np.nan <br>

### Operation on Null Values

**isnull()** <br>
Generate a Boolean mask indicating missing values<br>
**notnull()**<br>
Opposite of isnull()<br>
**dropna()**<br>
Return a filtered version of the data<br>
**fillna()**<br>
Return a copy of the data with missing values filled or imputed<br>

```python
data = pd.Series([1, np.nan, 'hello', None])
data
```

```python
df = pd.DataFrame([[1, np.nan, 2],
[2, 3, 5],
[np.nan, 4, 6]])
df
```

```python
#dropna() will drop all rows (observations) in which 
#any null value is present:
df.dropna()
```

```python
df.dropna(axis='columns') #or axis = 1 for variables
# will drop any column in which any null value is present
```

```python
df.dropna(axis=1)
```

```python
df[3] = np.nan  # column 3 set to NaN
df
```

```python
# default is "any null value is present"
# change to how = 'all'
df.dropna(axis='columns', how='all')
```

```python
df
```

```python
# if a row/column has too few non-null values, it should be removed
# thresh: if number of non-null values is less than threshold, it should be removed
df.dropna(axis='rows', thresh=3)
```
