## 1. 统计可视化

柱状图比较类别数值；堆叠柱状图把多组数值叠在一起，可加误差线。箱线图概括数值分布；Pandas 的 `DataFrame.plot` 可按列或指定 kind 画图，包括折线、柱状、直方图、箱线图和 KDE。

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({"A": [1, 2, 3, 4], "B": [2, 5, 3, 6], "C": [4, 3, 6, 5]})
df.plot(x="A", y="C", kind="line")
plt.show()
df.plot.box()
plt.show()
```

Seaborn 建立在 Matplotlib 之上，提供较高层次的统计图形。source 展示：
- `histplot`：查看单变量分布；
- `jointplot`：中心显示两个变量关系，边缘显示各自分布；
- `pairplot`：矩阵展示多个数值变量两两关系及单变量分布；
- Facet plots：按类别分面，比较不同子集；
- `catplot`：展示数值变量与一个或多个类别变量的关系。
这些图用于探索数据模式。

```python
import seaborn as sns
tips = sns.load_dataset("tips")
sns.pairplot(tips, hue="sex")
plt.show()
sns.jointplot(data=tips, x="total_bill", y="tip")
plt.show()
```

## 2. Scikit-learn 的基本步骤

source 将使用模型概括为：选择模型类；设置需要的超参数并创建实例；准备特征矩阵 X 和目标 y；调用 `fit` 学习；对新数据用 `predict` 预测，或用 `transform` 转换数据。监督学习提供目标标签，非监督学习的拟合不传入目标标签。

特征矩阵形状是 `[n_samples, n_features]`：每行一个样本、每列一个特征；单一目标通常是长度为 n_samples 的一维数组。简单线性回归示范一个特征，需将一维 x 整理为二维 X。

```python
import numpy as np
from sklearn.linear_model import LinearRegression

x = np.array([1, 2, 3, 4])
y = np.array([1, 3, 5, 7])
X = x[:, np.newaxis]

model = LinearRegression()
model.fit(X, y)
print(model.coef_, model.intercept_)
print(model.predict(np.array([[5], [6]])))
```

Scikit-learn 中的模型参数是 fit 后从训练数据估计出的属性；函数参数则是函数定义中接收调用值的变量。回归示例中的斜率和截距在模型拟合后可从 `coef_` 与 `intercept_` 查看。

## 4. 无监督学习示例

PCA 示例拟合特征 X（不传目标 y），再把数据转换为两个主成分，用散点图查看二维表示。KMeans 示例指定簇数，fit 后为样本预测簇标签，再按簇着色查看分组。

```python
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

pca = PCA(n_components=2)
X_2D = pca.fit_transform(X_iris)
clusters = KMeans(n_clusters=3, random_state=42).fit_predict(X_iris)
print(X_2D.shape, clusters[:10])
```

## 5. 三维图与三维等高线

source 还展示三维绘图：在三维 Axes 上用 `plot3D` 连接点，或用 `scatter3D` 绘制三维散点，并可将数值映射为颜色。三维等高线图 `contour3D` 使用网格坐标 X、Y 和对应的 Z 值绘制多个高度的等高线。

```python
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(projection="3d")
zline = np.linspace(0, 15, 100)
xline = np.sin(zline)
yline = np.cos(zline)
ax.plot3D(xline, yline, zline, color="gray")
plt.show()
```
