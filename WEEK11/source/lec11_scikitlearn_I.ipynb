### Supervised learning example: simple linear regression

```python
import matplotlib.pyplot as plt
import numpy as np
rng = np.random.default_rng(1)
x = 10 * rng.random(50)
y = 2 * x - 1 + rng.normal(loc=0, scale=1, size=50)
plt.scatter(x, y);
plt.show()
```

1. Choose a class of model.
In Scikit-Learn, every class of model is represented by a Python class. For simple linear regression model, import the linear regression class:

```python
from sklearn.linear_model import LinearRegression
```

2. Choose model hyperparameters. <br>
Once we have decided on our model class, there are still some options open to us. <br>
Depending on the model class we are working with, we might need to answer one or more questions.
In the LinearRegression model, hyperparameters are like the following: <br>
• fit_intercept=True:  Whether to calculate the intercept;If False, data is expected to be centered. <br>
• copy_X=True: If True, X is copied; otherwise it may be overwritten.<br>
• etc <br>

```python
LinearRegression?
```

```python
model = LinearRegression()
model
```

3. Arrange data into a features matrix and target vector.
two-dimensional features matrix and a one-dimensional target array

```python
y.shape  # y is in the correct form (a vector)
```

```python
x.shape  # x is not in the correct form (a matrix)
```

```python
# make x a matrix of size [n_samples (rows), n_features (columns)].
X = x[:, np.newaxis]
X.shape
```

4. Fit the data to your model.

```python
model.fit(X, y)
```

Check the slope and the intercept of the simple linear fit to the data:

```python
print(model.coef_)
print(model.intercept_)
```

```python
model?
```

5. Predict labels (yfit) for unknown data(Xfit). <br>
Coerce new x values into a [n_samples, n_features] features matrix, after which we can feed it to the model:

```python
xfit = np.linspace(-1, 11, num=20)   # 20 sample data
Xfit = xfit[:, np.newaxis]   # test data
Xfit.shape
```

```python
yfit = model.predict(Xfit)   # 50 predicted results are stored in yfit
```

```python
plt.scatter(x, y)
plt.scatter(xfit, yfit, color='red')
plt.show()
```

#### Feature Matrix
[n_samples, n_features]

```python
print(iris.iloc[:,0:-1].shape) #last column is excluded
```

```python
X_iris = iris.iloc[:,0:-1]
X_iris
```

```python
X_iris = iris.drop('species', axis=1)  #dataframe operations
X_iris 
```

#### Target Array
[n_samples, n_targets]

```python
y_iris = iris.iloc[:,-1] 
y_iris
```

```python
y_iris = iris['species']   #dataframe operations
y_iris
```

#### Data Exploration
Use plots effectively:

```python
%matplotlib inline
import seaborn as sns
sns.pairplot(iris, hue='species')  #different colors for different species
```

```python
from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain, ytest = train_test_split(X_iris, y_iris, random_state=1)
```

```python
print(Xtrain.shape)
print(Xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

```python
train_test_split?
```

### Choose model class
### We use GaussianNB class which is based on the Naive Bayes classification method

```python
from sklearn.naive_bayes import GaussianNB # 1. choose model class
model = GaussianNB() # 2. instantiate model
model.fit(Xtrain, ytrain) # 3. fit model to data
y_model = model.predict(Xtest) # 4. predict on new data
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest, y_model)  #compare the actual species with the predicted species
```

### 5-fold cross-validation
5-fold cross-validation is a standard technique for assessing a model's performance by splitting the data into five parts (folds). <br>
The model is trained on four folds and tested on the remaining one, and this process is repeated five times so every fold serves as the test set exactly once. <br>
You receive 5 distinct performance scores, which are typically averaged to give a reliable estimate of the model's true accuracy.

```python
from sklearn.model_selection import cross_val_score
model = GaussianNB() 
scores = cross_val_score(model, X_iris, y_iris, cv=5) # cv=5; 5-folds
print (scores.mean())
```

### Unsupervised learning example: dimension reduction

```python
from sklearn.decomposition import PCA # 1. Choose the model class 
# PCA belongs to linear dimensionality reduction 
model = PCA(n_components=2) # Reduces dataset dimensions to 2 principal components
model.fit(X_iris) # 3. Fit to data. Notice y is not specified!
```

```python
X_2D = model.transform(X_iris) # 4. Transform the data to two dimensions
```

```python
iris['PC1'] = X_2D[:, 0]
iris['PC2'] = X_2D[:, 1]
iris.head(5)
```

```python
sns.lmplot(data=iris, x="PC1", y="PC2", hue='species', fit_reg=False); # regression line isn't plotted
# plot a scatter plot for each category alongside a fitted linear regression line 
plt.show()
```

### Unsupervised learning example: clustering

```python
from sklearn.cluster import KMeans # 1. Choose the clustering model class (KMeans,GaussianMixture, etc.) 
model = KMeans(n_clusters=3, random_state=42) # 2. Instantiate the model w/ hyperparameters
model.fit(X_iris) # 3. Fit to data. Notice y is not specified!
```

```python
y_kmeans = model.predict(X_iris)  # 4. Determine cluster labels
```

```python
iris['cluster'] = y_kmeans  # categorical variable
sns.lmplot(x="PC1", y="PC2", data=iris, hue='species',
           col='cluster', fit_reg=False);  #Creates a grid of subplots (3 facets) by mapping 'cluster' across columns 
plt.show()
```
