## Visualization of Data
### Bar Charts

```python
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure()

# Add title and axis names
plt.title('Bar Chart of XX')
plt.xlabel('XX')
plt.ylabel('Count')

# Plot a bar chart
y = np.arange(1,11)  # y values are [1,....,9,10]
x = range(len(y))    # x values are [0,1,....,9]
plt.bar(x, y, width = 0.3);
```

```python
#  Bar chart with error bar
# plot errorbars using the list of random error values
err = np.random.random(len(y))
plt.bar(x, y, width = 0.3, yerr=err);
```

### Two bar charts on one axes

```python

y1 = np.arange(1,11)         # y values are [1,....,9,10]
x1 = range(len(y1))          # x values are [0,1,....,9]
plt.bar(x1, y1, width = 0.3);

y2 = y1 ** 2      # plot another set of bars
# adjusting the new x values for the second set of bars 
# x2 = x1 + 0.3      not working; TypeError: unsupported operand type(s) for +: 'range' and 'float'
x2 = []
for item in x1:
    x2.append(item+0.3)

plt.bar(x2, y2, width = 0.3 ,color='lightgreen');
```

```python
# stacked bar charts 
plt.figure()

plt.bar(x1, y1, width = 0.3, color='b')
plt.bar(x1, y2, width = 0.3, 
        bottom=y1, #stacked
        color='lightgreen');
```

```python
# barh: horizontal bar charts

plt.barh(x1, y1, height = 0.3, color='b')
plt.barh(x1, y2, height = 0.3, 
         left=y1, 
         color='lightgreen');
```

### DataFrame.plot
it plots every numeric column against the DataFrame's index.

```python
np.random.seed(1)
#create a df with three 1-D arrays which have 365 std normal distribution cumulative random values
df = pd.DataFrame({'A': np.random.randn(365).cumsum(),      
                   'B': np.random.randn(365).cumsum() + 20,
                   'C': np.random.randn(365).cumsum() - 20}, 
                  index=pd.date_range('11/11/2024', periods=365)) # a date range as the row index
df.head()
```

```python
# add a semi-colon to suppress the output
df.plot();
plt.show()  # optional, use if output is blocked
```

We can select which plot we want to use by passing it into the 'kind' parameter. <br>
Alternatively, use `DataFrame.plot.kind` methods.

`kind` :
- `'line'` : line plot (default)
- `'bar'` : vertical bar plot
- `'barh'` : horizontal bar plot
- `'hist'` : histogram
- `'box'` : boxplot
- `'kde'` : Kernel Density Estimation plot
- `'density'` : same as 'kde'
- `'area'` : area plot
- `'pie'` : pie plot
- `'scatter'` : scatter plot
- `'hexbin'` : hexbin plot

```python
df.plot('A','C', kind = 'scatter');
```

```python
# scatterplot of 'A' and 'C'
# color (c) and size (s) is marker size
df.plot.scatter('A', 'C', c='B', s= 20 , colormap='viridis');
# color act as the third dimension;a colormap act as the 3rd dimension axis
```

```python
df.plot.box();
```

```python
df.plot.hist(bins=20,alpha=0.3);
```

```python
# kernel density estimation plot instead 
df.plot.kde();
```

```python
# hexagonal binning plot
df.plot('A','C', kind = 'hexbin');
# the colors of the hexagons act as the third dimension; 
# each hexagon is coloured by how many points fall inside it (dark = many points).
# densities of points are also viewed
```

### Pnadas Scatter_matrix()
- generate a grid of scatter plots for every pair of numeric columns in a DataFrame.
- helps you visually inspect relationships and distributions across multiple variables at once.
- Diagonals: The plots on the diagonal (where a variable is plotted against itself) typically show the distribution of that variable, using a histogram or a density plot.
- Off-Diagonals: The off-diagonal plots are scatter plots that show the relationship between each pair of variables.

```python
pd.plotting.scatter_matrix(df, alpha=0.2);  #alpha: Transparency of the points
```

## Seaborn
Seaborn is a dedicated statistical visualization library designed to create more complex, informative, and aesthetically pleasing graphics with less code (high-level commands). <br>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

```python
# Create a data array contains two arrays of correlated data
# Bivariate data : Pairs of observations recorded for the same subjects or cases 
# (e.g., measuring both height and weight for each person in a group).

data_array = np.random.multivariate_normal([0, 30], # [mean of group x ,  mean of group y]
                                     [[100, 50],   # Covariance matrix  [var(x)     covar(x,y)]
                                      [50, 100]],  # (symmetric)        [covar(y,x)  var(y)   ]
                                     size=2000)    # Number of samples
data_df = pd.DataFrame(data_array, columns=['x', 'y'])
```

### Compare the 3 ways to plot histograms
- Matplotlib: plt.hist()
- Pandas: pd.plot.hist()
- Seaborn: sns.histplot()

```python
# Matplotlib plots
""" 
Better to use loop if many columns in the dataframe
for col in 'xy':
    plt.hist(data_df[col], density=True, alpha=0.5,
             bins=20, label = col) 
"""
plt.hist(data_df['x'], density=True, alpha=0.5,bins=20, label = 'x')
plt.hist(data_df['y'], density=True, alpha=0.5,bins=20, label = 'y')
plt.legend();
```

```python
# Pandas dataframe plots
data_df.plot.hist(bins=20,alpha=0.3);
```

```python
# Seaborn plots
sns.histplot(data=data_df, x='x', kde=True)  # can add kde curve on the histogram 
sns.histplot(data=data_df, x='y', kde=True);
```

### Seaborn Jointplot()
Creates a single figure combining a bivariate plot (showing the relationship between two variables) in the center with univariate plots (showing each variable's distribution) on the margins. 

```python
sns.jointplot( data=data_df, x="x",y="y", alpha=0.4, kind='hex');
```

```python
sns.jointplot( data=data_df, x="x",y="y", alpha=0.4, kind='kde');
```

```python
sns.jointplot( data=data_df, x="x",y="y", kind='reg');   #regression
```

### Seaborn Pairwise plot
- designed for exploratory data analysis.
- creates a matrix of plots that allows you to quickly visualize
    - the pairwise relationships between all numeric variables in a dataset
    - the univariate distribution of each individual variable.

```python
np.random.seed(1234)
df = pd.DataFrame({'A': np.random.randn(100).cumsum(), 
                   'B': np.random.randn(100).cumsum() + 20,
                   'C': np.random.randn(100).cumsum() - 20, 
                   'D': np.random.randint(0,3,100)})            # as a group index

sns.pairplot(df, hue='D',palette='Accent',diag_kind='kde'); 

# a grid of scatter plots it helps to identify how different features interact with each other
# diagonal plots are distributions of each variable
# hue: a Seaborn method's argument to specify different colors to represent different data groups (group 0, 1 and 2 ).
```

### Seaborn Facet Plots
- facet plots are a way to create "small multiples".
- let you draw the same type of plot on different subsets of your dataframe, arranging them in a grid.
- makes it easy to compare patterns across different categories at a glance.

```python
# the amount that restaurant staff receive in tips 
# based on various indicator data
tips_df = sns.load_dataset('tips')
tips_df['tip_pct'] = 100 * tips_df['tip'] / tips_df['total_bill']
tips_df.head()
```

```python
# Multiple facets to show the tips relationships in the tips_df dataframe 
grid = sns.FacetGrid(tips_df, row="sex", col="time",  # create and control the layout and appearance of your grid
                     margin_titles=True)      # split the data into  groups according  to the "sex" and "time"

# map a plotting function: Apply a plotting function (like plt.hist) to each facet in the grid using the .map() 
grid.map(plt.hist,     # plotting function name in each facet
         "tip_pct", bins=np.linspace(0, 40, 10));  # arguments passed to the plotting function
plt.show()
```

####  Seaborn Catplot
- creating categorical plots.
- can draw a variety of plot types (specify by 'kind' argument)
- show the relationship between a numerical variable and one or more categorical variables.

```python
# The catplot() function provides access to several axes-level functions that
# show the relationship between a numerical and one or more categorical
# variables using one of several visual representations.

grid = sns.catplot(data=tips_df,
                x="day", y="total_bill",  #show how "total_bill" depends on "day"
                hue="sex",    # divide data into sex groups
                kind="box")   # box plot to show the relationship
grid.set_axis_labels("Day", "Total Bill");
```

```python
grid = sns.catplot(data=tips_df,
                x="day", y="total_bill",  #show how "total_bill" depends on "day"
                hue="sex",    # divide data into sex groups
                kind="violin")   # box plot to show the relationship
grid.set_axis_labels("Day", "Total Bill");
```

```python
tips_df.describe()   # show the 'total_bill' distribution by statistical values
```

```python
# violin plot to show the 'total_bill' distribution for each 'day'
sns.violinplot(x="day", y="tip", hue="sex", data=tips_df,
                   split=True, inner="quartile",
                   palette=["lightblue", "lightpink"]);
```

#### 3D plots

```python
from mpl_toolkits import mplot3d
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
fig = plt.figure()
ax = plt.axes(projection='3d')
```

```python
ax = plt.axes(projection='3d')
# Data for a three-dimensional line
zline = np.linspace(0, 15, 1000)
xline = np.sin(zline)
yline = np.cos(zline)
ax.plot3D(xline, yline, zline, 'gray')
# Data for three-dimensional scattered points
zdata = 15 * np.random.random(100)
xdata = np.sin(zdata) + 0.1 * np.random.randn(100)
ydata = np.cos(zdata) + 0.1 * np.random.randn(100)
ax.scatter3D(xdata, ydata, zdata, c=zdata, cmap='Reds'); #points with transparency adjusted
plt.show()
```

#### Three-Dimensional Contour Plots

```python
def f(x, y):
    return np.sin(np.sqrt(x ** 2 + y ** 2))
x = np.linspace(-6, 6, 30)
y = np.linspace(-6, 6, 30)
X, Y = np.meshgrid(x, y)  #convert two 1-D arrays to become two 2-D arrays
Z = f(X, Y)
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')
ax.contour3D(X, Y, Z, 50, cmap='binary')   # 50 contour lines
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
plt.show()
```
